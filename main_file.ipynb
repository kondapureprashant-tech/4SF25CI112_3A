Handling the Missing Value

```python
df.isnull().any()
```

```python
df.isnull().sum()
```

```python
df_filled=df.fillna(0)
```

```python
df_filled
```

```python
df =df.rename(columns= {'Series_title_1':'SeriesVolume'})
df.head()
```

```python
df['city']=df['cgpa'].fillna(df['cgpa'].mean()).astype(int)
df.head()
```

```python
grouped_mean =df.groupby('internal_marks')['assignment_score'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby(['final_exam_marks','sleep_hours'])['cgpa'].sum()
print(grouped_sum)
```

```python
grouped_agg = df.groupby('internet_hours')['cgpa'].agg(['mean','sum','count'])
grouped_agg
```


Dropping Unnecessary Columns

```python

df = df.drop(['student_id', 'age', 'city'], axis=1)
```

```python
df
```


Mean Value Imputation

```python

import seaborn as meta
```

```python
meta.histplot(df['gender'],kde=True)
```

```python
df['internal_marks'] = df['internal_marks'].fillna(df['internal_marks'].mean())
df[['internal_marks','assignment_score']].head(10)
```

```python
meta.histplot(df['internal_marks'],kde=True)
```

```python
df['monthly_expense_inr'] =df['internet_hours'].fillna(df['internet_hours'].mean()) 
df[['monthly_expense_inr','internet_hours']]
```


Insight Generation


The Top Export Growth Products

```python
yearly_trend = df.groupby('assignment_score')['internal_marks'].mean()

yearly_trend.head(5)
yearly_trend.tail(5)
```

```python
volatility = df.groupby('department')['study_hours_per_day'].std().sort_values(ascending=False)

volatility.head(3)
```

```python
status_breakdown = df['study_hours_per_day'].value_counts()
status_breakdown
```

```python
trend = df.groupby('sleep_hours')['cgpa'].mean()

trend.plot(kind='line', title='Export Trends Over Time')
```

```python
top_products = df.groupby('study_hours_per_day')['attendance_percent'].mean().sort_values(ascending=False).head(5)

top_products.plot(kind='bar', title='Top 5 Export Products')
```

```python
status_counts = df['department'].value_counts()

status_counts.plot(kind='bar', title='Data Status Breakdown')
```
